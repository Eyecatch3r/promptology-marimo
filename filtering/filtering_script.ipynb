# Imports

```python
import pandas as pd
import numpy as np
```

# Data

```python
df_corpus = pd.read_csv('dummy_data/dummy_corpus.csv', delimiter=';') # Adapt as needed
df_comb = pd.read_csv('dummy_data/dummy_combinations.csv', delimiter=';') # Adapt as needed
```

# Filtering

```python
rows_de = []
rows_it = []
rows_fr = []
for index, row in df_comb.iterrows():
    if not df_corpus[(df_corpus['language'] == 'de')].empty:
        df_de_filter = df_corpus[(df_corpus['language'] == 'de') & (df_corpus['ID_question'] == row['ID_question'])]
        if not df_de_filter[df_de_filter[row['socio_dem']] == row['attribute']].empty:
            rows_de.append(row)

    if not df_corpus[(df_corpus['language'] == 'it')].empty:
        df_it_filter = df_corpus[(df_corpus['language'] == 'it') & (df_corpus['ID_question'] == row['ID_question'])]
        if not df_it_filter[df_it_filter[row['socio_dem']] == row['attribute']].empty:
            rows_it.append(row)

    if not df_corpus[(df_corpus['language'] == 'fr')].empty:
        df_fr_filter = df_corpus[(df_corpus['language'] == 'fr') & (df_corpus['ID_question'] == row['ID_question'])]
        if not df_fr_filter[df_fr_filter[row['socio_dem']] == row['attribute']].empty:
            rows_fr.append(row)

df_de_comb = pd.DataFrame(rows_de)
df_it_comb = pd.DataFrame(rows_it)
df_fr_comb = pd.DataFrame(rows_fr)
```

# Export

```python
if not df_de_comb.empty:
    df_de_comb.to_csv('filtered_comb/de_combination.csv', index=False) # Adapt as needed

if not df_it_comb.empty:
    df_it_comb.to_csv('filtered_comb/it_combination.csv', index=False) # Adapt as needed

if not df_fr_comb.empty:
    df_fr_comb.to_csv('filtered_comb/fr_combination.csv', index=False) # Adapt as needed
```
