```python
import pandas as pd
```

```python
def unroll_choices(df):
    df["stop_reason"] = df["choices"].apply(lambda x: x[0]["finish_reason"])
    df["output"] = df["choices"].apply(lambda x: x[0]["message"]["content"])

    return df

def preprocess(df):
    df = unroll_choices(df)
    df["output_id"] = df["id"]
    df["model"] = "gpt-4.1-mini"
    df = df[["prompt_id", "output_id", "output", "stop_reason", "model"]]
    return df
```

# German

```python
df = pd.read_json("output/gpt-4.1-mini_de.jsonl", lines=True)
```

```python
df = preprocess(df)
```

```python
df.to_csv("output/gpt-4.1-mini_de.csv", index=False)
```

# French

```python
df = pd.read_json("output/gpt-4.1-mini_fr.jsonl", lines=True)
df = preprocess(df)
df.to_csv("output/gpt-4.1-mini_fr.csv", index=False)
```
