```python
import json
import pandas as pd
from transformers import AutoTokenizer
```

```python
prompts_de = pd.read_csv("prompts_de.csv")
prompts_it = pd.read_csv("prompts_it.csv")
prompts_fr = pd.read_csv("prompts_fr.csv")
```

```python
outputs = pd.read_json("output/llama3.1-8B-instruct_de.jsonl", lines=True)

batch_size = 32

ids = []
for i in range(0, len(prompts_de), batch_size):
    for j in range(3):
        for k in range(batch_size):
            if i + k < len(prompts_de):
                ids.append(prompts_de["id"][i + k])
outputs["id"] = ids
```

```python
outputs.to_csv("output/llama3.1-8B-instruct_de.csv", index=False)
```

```python
outputs.to_json("output/llama3.1-8B-instruct_de.jsonl", lines=True, orient="records")
```
